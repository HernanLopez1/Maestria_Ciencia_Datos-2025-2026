<h3 id="#identificacion-columnas-nulos">2.4 Identificación de columnas con valores faltantes (NaN - None)<h3

```python
def mostrar_informacion_nulos(df_concatenado):
    nulos = df_concatenado.isnull().sum()
    porcentaje_nulos = df_concatenado.isnull().mean() * 100
 
    resumen_nulos = pd.DataFrame({
        'Conteo nulos': nulos,
        'Porcentaje nulos (%)': porcentaje_nulos.round(2)
    })
 
    # Ordenar las columnas por porcentaje de nulos descendente
    resumen_nulos = resumen_nulos.sort_values('Porcentaje nulos (%)', ascending=False)
 
    display(Markdown("**Conteo y porcentaje de valores nulos por columna (ordenado de mayor a menor)**"))
    print(resumen_nulos.to_markdown())
   
mostrar_informacion_nulos(df_concatenado)
```

```python
def convertir_fecha_espanol(fecha_str):
    meses_espanol = {
        'ENE': 'Jan', 'FEB': 'Feb', 'MAR': 'Mar', 'ABR': 'Apr',
        'MAY': 'May', 'JUN': 'Jun', 'JUL': 'Jul', 'AGO': 'Aug',
        'SEP': 'Sep', 'OCT': 'Oct', 'NOV': 'Nov', 'DIC': 'Dec'
    }

    if pd.isna(fecha_str):
        return pd.NaT

    try:
        partes = fecha_str.split('-')
        if len(partes) == 3:
            dia, mes, anio = partes
            mes_ingles = meses_espanol.get(mes.upper(), mes)
            nueva_fecha = f"{dia}-{mes_ingles}-{anio}"
            return pd.to_datetime(nueva_fecha, format='%d-%b-%Y')
        return pd.NaT
    except:
        return pd.NaT

# Aplicar la función
df_final['FECHA_ICFES'] = df_final['FECHA_ICFES'].apply(convertir_fecha_espanol)
df_final['FECHA_NACIMIENTO'] = df_final['FECHA_NACIMIENTO'].apply(convertir_fecha_espanol)
```

```python
# Extraer el año de la fecha de nacimiento
df_final['Año_Nacimiento'] = df_final['FECHA_NACIMIENTO'].dt.year

df_final['Edad_Pro'] = df_final['Año'] - df_final['Año_Nacimiento']

df_final['Edad_Pro'] = df_final['Edad_Pro'].fillna(0).astype(int)
df_final['Año_Nacimiento'] = df_final['Año_Nacimiento'].fillna(0).astype(int)

df_final['Año_Ingreso_Estudiante'] = df_final['PERIODO_INGRESO_ESTUDIANTE'].astype(str).str[:4].astype(int)
 # Calcular la diferencia en años (Año del examen - Año de ingreso del estudiante)
df_final['Tiempo_Estudio'] = df_final['Año'] - df_final['Año_Ingreso_Estudiante']

# Extraer el año de la fecha del ICFES
df_final['Año_ICFES'] = df_final['FECHA_ICFES'].dt.year

# Calcular la diferencia en años entre el año del examen Saber Pro y el año del ICFES
df_final['Diferencia_Años_ICFES_SaberPro'] = df_final['Año'] - df_final['Año_ICFES']

# Convertir a Int64 (que permite valores nulos)
df_final['Año_ICFES'] = df_final['Año_ICFES'].astype('Int64')
df_final['Diferencia_Años_ICFES_SaberPro'] = df_final['Diferencia_Años_ICFES_SaberPro'].astype('Int64')


print("DataFrame con las nuevas columnas':")
display(df_final[['Año_Nacimiento', 'Edad_Pro', 'PERIODO_INGRESO_ESTUDIANTE', 'Año_Ingreso_Estudiante', 'Tiempo_Estudio', 'Año_ICFES', 'Diferencia_Años_ICFES_SaberPro']].head())
```

<h2 id="#Tratamiento de Outliers">6.2 Tratamiento de Outliers</h2>

##### Método basado en el rango intercuartílico (IQR)


**Nota:** 

En el análisis univariado, a través del boxplot se identificó que algunas variables numéricas, como el Puntaje global, presentan valores atípicos. Por lo que, se emplea el método del rango intercuartílico (IQR) para su detección y posterior eliminación. 

```python
def detectar_atipicos_simple(df, umbral=1.5):
    """
    Detección simple de atípicos usando método IQR
    """
    columnas_numericas = df.select_dtypes(include=[np.number]).columns
    resultados = []
    
    for col in columnas_numericas:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lim_inf = Q1 - umbral * IQR
        lim_sup = Q3 + umbral * IQR
        
        atipicos = df[(df[col] < lim_inf) | (df[col] > lim_sup)]
        total_atipicos = len(atipicos)
        porcentaje = (total_atipicos / len(df)) * 100
        
        resultados.append({
            'columna': col,
            'total_atipicos': total_atipicos,
            'porcentaje_atipicos': round(porcentaje, 2),
            'limite_inferior': lim_inf,
            'limite_superior': lim_sup,
            'minimo': df[col].min(),
            'maximo': df[col].max()
        })
    
    return pd.DataFrame(resultados)

# Uso rápido
detectar_atipicos_simple(numericas)
```

```python
def eliminar_atipicos(df, umbral=1.5):
    columnas_numericas = df.select_dtypes(include=[np.number]).columns
    df_sin_outliers = df.copy()

    for col in columnas_numericas:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lim_inf = Q1 - umbral * IQR
        lim_sup = Q3 + umbral * IQR
        
        df_sin_outliers = df_sin_outliers[(df_sin_outliers[col] >= lim_inf) & (df_sin_outliers[col] <= lim_sup)]

    return df_sin_outliers
numericas_sin_outliers = eliminar_atipicos(numericas, umbral=1.5)
```

**Interpretación:**

Teniendo en cuenta lo anterior, se considera un puntaje atipico aquellos valores menores al limite inferior y mayores al limite superior de cada una de las variables numéricas. Este criterio se fundamenta en la dispersión propia de la mayoría de los resultados de los estudiantes, de modo que los valores que se encuentran fuera de esos rangos se consideran inusuales.

<h3 id="Modelo N° 1">8.1 Modelo N° 1</h3>

##### Se definen las variables y se aplica One-Hot Encoding para la variables categoricas

```python
# Definir la variable objetivo (y)
y = df_final['Puntaje global']
 
# Lista de variables predictoras (X) para el Modelo 1
columnas_x_originales = [
    'PUNTAJE_CIENCIAS_NATURALES_SABER_11',
    'PUNTAJE_COMPETENCIAS_CIUDADANAS_SABER_11',
    'PUNTAJE_MATEMATICAS_SABER_11',
    'PUNTAJE_LECTURA_CRITICA_SABER_11',
    'PUNTAJE_INGLES_SABER_11',
    'PROMEDIO_ACUMULADO',
    'CIUDAD_NACIMIENTO',
    'CIUDAD_COLEGIO',
 
]
 
# Seleccionar las columnas predictoras originales del DataFrame
X = df_final[columnas_x_originales].copy()
 
# Identificar las columnas categóricas para el Modelo 1
columnas_categoricas_x = X.select_dtypes(include=['object', 'category']).columns.tolist()
 
# Columnas categóricas que tendrán encoding binario específico para 'Cali'
columnas_ciudad_cali = ['CIUDAD_NACIMIENTO', 'CIUDAD_COLEGIO']
 
# Columnas categóricas que tendrán One-Hot Encoding estándar
columnas_ohe_estandar = [col for col in columnas_categoricas_x if col not in columnas_ciudad_cali]
 
# Aplicar encoding binario para las columnas de ciudad (Cali vs No-Cali)
for col in columnas_ciudad_cali:
    if col in X.columns:
        X[f'{col}_Cali'] = (X[col] == 'Cali').astype(int)
        # Opcional: Eliminar la columna original después de crear la binaria
        X = X.drop(columns=[col])
 
# Aplicar One-Hot Encoding estándar a las otras columnas categóricas
if columnas_ohe_estandar:
    X = pd.get_dummies(X, columns=columnas_ohe_estandar, drop_first=True)
 
print("\n--- Tipos de datos del DataFrame X después del encoding ---")
print(X.dtypes)
print("--------------------------------------------------------")
```

##### Se dividen los datos en conjuntos de entrenamiento y prueba

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
 
print("Dimensiones de X_train:", X_train.shape)
print("Dimensiones de X_test:", X_test.shape)
print("Dimensiones de y_train:", y_train.shape)
print("Dimensiones de y_test:", y_test.shape)
 
 # Inicializar el modelo de Regresión Lineal
linear_model = LinearRegression()
 
# Entrenar el modelo con los datos de entrenamiento
linear_model.fit(X_train, y_train)
 
y_pred = linear_model.predict(X_test)
 
print("Modelo de Regresión Lineal Múltiple entrenado exitosamente.")
# Puedes ver los coeficientes del modelo después de entrenar
# print("Coeficientes del modelo:", linear_model.coef_)
# print("Intercepto del modelo:", linear_model.intercept_)
```

```python
# Definir la variable objetivo (y) para el Modelo 4
y4 = df_final['Puntaje global']

# Lista de variables predictoras (X) para el Modelo 4
columnas_x4_originales = [
    'PUNTAJE_GLOBAL_SABER_11',
    'PUNTAJE_CIENCIAS_NATURALES_SABER_11',
    'PUNTAJE_COMPETENCIAS_CIUDADANAS_SABER_11',
    'PUNTAJE_MATEMATICAS_SABER_11',
    'PUNTAJE_LECTURA_CRITICA_SABER_11',
    'PUNTAJE_INGLES_SABER_11',
    'PROMEDIO_ACUMULADO',
    'CIUDAD_NACIMIENTO', # Binaria específica (Cali)
    'CIUDAD_COLEGIO', # Binaria específica (Cali)
    'DEPTO_COLEGIO', # Binaria específica (Valle del cauca)
    'SER_PILO_PAGA', # Binaria (0/1)
    'SEXO', # Binaria (0/1)
    'Año_Nacimiento',
    'NACIONALIDAD', # Binaria específica (colombiana)
    'Año_Ingreso_Estudiante',
    'Año_ICFES',
    'Año',
    'ESTRATO', # Numérica
    'Diferencia_Años_ICFES_SaberPro',
    'Tiempo_Estudio',
    'Edad_Pro',
]


columnas_x4_existentes = [col for col in columnas_x4_originales if col in df_final.columns]
X4 = df_final[columnas_x4_existentes].copy()

# Identificar columnas categóricas originales seleccionadas para Modelo 4
columnas_categoricas_x4 = X4.select_dtypes(include=['object', 'category']).columns.tolist()

# Columnas que tendrán encoding binario específico ('Cali', 'Valle del cauca', 'colombiana')
columnas_binarias_especificas = ['CIUDAD_NACIMIENTO', 'CIUDAD_COLEGIO', 'CIUDAD_RESIDENCIA', 'DEPTO_COLEGIO', 'NACIONALIDAD']


columnas_binarias_mapeo = ['SEXO', 'SER_PILO_PAGA'] 

# Columnas nominales que quedan para OHE estándar
columnas_nominales_ohe_estandar = [col for col in columnas_categoricas_x4 if col not in columnas_binarias_especificas + columnas_binarias_mapeo]


# Aplicar Codificación Binaria Específica
for col in columnas_binarias_especificas:
    if col in X4.columns:
        if col in ['CIUDAD_NACIMIENTO', 'CIUDAD_COLEGIO', 'CIUDAD_RESIDENCIA']:
      
            X4[f'{col}_Cali'] = (X4[col] == 'Cali').astype(int)
        elif col == 'DEPTO_COLEGIO':
          
            X4[f'{col}_ValleDelCauca'] = (X4[col] == 'VALLE DEL CAUCA').astype(int)
        elif col == 'NACIONALIDAD':
       
            X4[f'{col}_Colombiana'] = (X4[col] == 'CO').astype(int)
        # Eliminar la columna original después de crear la binaria
        X4 = X4.drop(columns=[col])

# Aplicar Codificación Binaria por Mapeo (0/1)
for col in columnas_binarias_mapeo:
    if col in X4.columns:
         if col == 'SEXO':
             # Mapeo M a 0, F a 1
             X4[col] = X4[col].map({'M': 0, 'F': 1}).astype(int)
         elif col == 'SER_PILO_PAGA':
             # Mapeo N a 0, S a 1
             X4[col] = X4[col].map({'N': 0, 'S': 1}).astype(int)


# Aplicar One-Hot Encoding estándar a las columnas nominales restantes
if columnas_nominales_ohe_estandar:
    X4 = pd.get_dummies(X4, columns=columnas_nominales_ohe_estandar, drop_first=True)


print("Variables predictoras (X4) para el Modelo 4 después de selección y encoding:")
display(X4.head())
print("\nForma de las variables predictoras (X4):", X4.shape)
print("\nVariable objetivo (y4):")
display(y4.head())
print("\nForma de la variable objetivo (y4):", y4.shape)
```

##### Se entrena y se ajusta el modelo

```python
# Agregar una constante a las variables predictoras (necesario para statsmodels OLS)
X4_train_sm = sm.add_constant(X4_train, prepend=True)


X4_train_sm.columns = ['const'] + list(X4_train.columns)

# Inicializar y ajustar el modelo de Regresión Lineal Múltiple (OLS) usando statsmodels para el Modelo 4
model4 = sm.OLS(y4_train, X4_train_sm)
result4 = model4.fit()

print("Modelo de Regresión Lineal Múltiple (Modelo 4) entrenado exitosamente usando statsmodels.")

# Imprimir el resumen del modelo
print(result4.summary())
```
